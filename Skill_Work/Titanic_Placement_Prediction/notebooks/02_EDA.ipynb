```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("/Users/kasturi/Desktop/Machine_Learning_-TRI_SEM_4-/Skill_Work/Titanic_Placement_Prediction/data/titanic.csv")
```

```python
df.head()
```

```python
# Numerical vs Categorical features
numerical_features = df.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_features = df.select_dtypes(
    include=['object']
).columns

print("Numerical Features:", list(numerical_features))
print("Categorical Features:", list(categorical_features))
```

```python
numerical_cols = df.select_dtypes(
    include=[np.number]
).columns.tolist()

corr_matrix = df[numerical_cols].corr()
```

```python
print("\n--- Correlation Matrix ---")
print(corr_matrix)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,15))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5
)

plt.title(
    "Correlation Heatmap of Numerical Features",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
target_col = "PlacementStatus"
```

```python
target_col = "Survived"

for col in numerical_cols:

    if col != target_col:

        plt.figure(figsize=(6, 5))

        sns.boxplot(
            x=target_col,
            y=col,
            data=df
        )

        plt.title(f"{col} vs {target_col}")
        plt.xlabel("Survived")
        plt.ylabel(col)

        plt.tight_layout()
        plt.show()
```

```python
numerical_columns = [
    "Age",
    "Fare",
    "SibSp",
    "Parch",
    "Pclass"
]

fig6 = plt.figure(figsize=(14, 8))

df[numerical_columns].boxplot()

plt.xticks(rotation=45)
plt.title("Titanic Numerical Features - Boxplots")

plt.show()
```

```python
fig7 = sns.boxplot(
    x='Survived',
    y='Fare',
    data=df
)

plt.title("Fare vs Survival")
plt.xlabel("Survived")
plt.ylabel("Fare")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Sex",
    hue="Survived"
)

plt.title("Survival by Sex")
plt.xlabel("Sex")
plt.ylabel("Number of Passengers")

plt.show()
```
